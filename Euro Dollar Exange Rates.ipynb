# Euro–Dollar Exchange Rates: Data Storytelling (1999–2021)

The dataset describes Euro daily exchange rates between 1999 and 2021. The euro is the official currency in most of the countries of the European Union.he data source is the European Central Bank.


```python
import pandas as pd

exchange_rates = pd.read_csv('euro-daily-hist_1999_2020.csv')
exchange_rates.head()
```

```python
exchange_rates.tail()
```

```python
exchange_rates.info()
```

## Data Cleaning

```python
#renaming [US dollar ] and Period\Unit: columns 
#changing the Time column to a datetime data type
#sorting the values by Time in ascending order
#reseting the index (and drop the initial index)

exchange_rates.rename(columns={'[US dollar ]': 'US_dollar',
                               'Period\\Unit:': 'Time'},
                      inplace=True)
exchange_rates['Time'] = pd.to_datetime(exchange_rates['Time'])
exchange_rates.sort_values('Time', inplace=True)
exchange_rates.reset_index(drop=True, inplace=True)
exchange_rates.head()
```

```python
euro_to_dollar = exchange_rates[['Time', 'US_dollar']].copy()
euro_to_dollar['US_dollar'].value_counts() # 62 '-' characters
```

### Converting the US_dollar column to a float data type and excluding all the rows where the '-' character appears in the US_dollar column

```python
euro_to_dollar = euro_to_dollar[euro_to_dollar['US_dollar'] != '-']
euro_to_dollar['US_dollar'] = euro_to_dollar['US_dollar'].astype(float)
euro_to_dollar.info()
```

## Rolling Mean

Rolling mean is calculated using values from two days — the current day and the previous day. We call the number of days used in the calculation the rolling window (or moving window)

```python
import matplotlib.pyplot as plt
%matplotlib inline

#generating a line plot to visualize the evolution of the euro-dollar exchange rate.

plt.plot(euro_to_dollar['Time'], euro_to_dollar['US_dollar'])
plt.show()
```

## Euro-dollar exchange rate

```python
#adding the rolling means to a new column, a 30 days moving window was used
euro_to_dollar['rolling_mean'] = euro_to_dollar['US_dollar'].rolling(30).mean()
```

```python
#generating a line plot
fig,ax = plt.subplots(figsize=(10,5))
plt.plot(euro_to_dollar['Time'],
         euro_to_dollar['rolling_mean'])
plt.title('EURO - USD rate change between 1999 and 2021',size =16)
plt.xlabel('Year',size = 12)
plt.ylabel('Rate',size = 12)
plt.show()
```

### Euro-Dollar over time

```python
plt.figure(figsize=(9,6))

plt.subplot(3,2,1)
plt.plot(euro_to_dollar['Time'], euro_to_dollar['US_dollar'])
plt.title('Original values', weight='bold')

for i, rolling_mean in zip([2, 3, 4, 5, 6],
                           [7, 30, 50, 100, 365]):
    plt.subplot(3,2,i)
    plt.plot(euro_to_dollar['Time'],
             euro_to_dollar['US_dollar'].rolling(rolling_mean).mean())
    plt.title('Rolling Window:' + str(rolling_mean), weight='bold')
    
plt.tight_layout() # Auto-adjusts the padding between subplots
plt.show()
```

## Euro-dollar rate change during Coronavirus Pandemic

Let's have closer look to the more specific period of time and explore how the euro-dollar rate has changed during the Coronavirus Pandemic. The 2020 data will be shown and the 2016-2019 data will be used as a baseline. A line plot will represent the data set.

```python
#preparing the required time frame
pandemic_euro = euro_to_dollar.copy()[
    (euro_to_dollar['Time'].dt.year >= 2016) & 
    (euro_to_dollar['Time'].dt.year <= 2021)]

pandemic_euro_2020 = euro_to_dollar.copy()[
    (euro_to_dollar['Time'].dt.year >= 2020) & 
    (euro_to_dollar['Time'].dt.year <= 2021)]

#calculating the rolling mean
round(pandemic_euro_2020['rolling_mean'].min(),2)
```

```python
import matplotlib.style as style
style.use('fivethirtyeight')

#generating a line plot
fig,ax = plt.subplots(figsize=(10,5))
ax.plot(pandemic_euro['Time'],
         pandemic_euro['rolling_mean'],lw=1, color= 'blue')

# adding the highlight part for pandemic period
ax.plot(pandemic_euro_2020['Time'],
         pandemic_euro_2020['rolling_mean'],lw= 2.5, color= 'red')

#changing the color of tick labels
ax.tick_params(axis ='x', colors = 'grey')
ax.tick_params(axis ='y', colors = 'grey')

# removing labels on both axes
ax.set_xticklabels([])
ax.set_yticklabels([])

#to assign new labels we need to find out the tic's coordinates.
#the follwing functions were used:
#ax.get_xticks()
#the output: array([16436. 16801. 17167. 17532. 17897. 18262. 18628])
#ax.get_yticks()
#the output: array([1.025 1.05  1.075 1.1   1.125 1.15  1.175 1.2   1.225 1.25])

#assigning new labes

x = 16701
for year in  ['2016', '2017', '2018', '2019', '2020', '2021']:
    ax.text(x, 1.03, year, alpha=0.5, fontsize=15)
    x += 365
    
y = 1.05
for rate in ['1.05', '1.1', '1.15', '1.2','1.25']:
    ax.text(16600, y, rate, alpha=0.5, fontsize=15)
    y += 0.05

# adding title and labels
ax.text(16801,1.28,'EURO - USD rate change during 2016-2021 period of time',
        weight = 'bold', size=20, rotation = 0)
ax.text(18262,1.25,'Coronavirus Pandemic',
        weight = 'bold', size=15, rotation = 0)
ax.text(16540,1.15,'Rate',color ='black',alpha=0.3, size=15, rotation = 90)
ax.text(17532,1.01,'Year',color ='black',alpha=0.3, size=15, rotation = 0)

# highlighting the period of pandemic
ax.axvspan(xmin=18262, xmax=18650, ymin=0.03,
           alpha=0.3, color='grey')

plt.show()

# # print(ax.get_xticks())
# # print(ax.get_yticks())
```

#### The Three US Presidencies Example

```python
bush_obama_trump = euro_to_dollar.copy(
                   )[(euro_to_dollar['Time'].dt.year >= 2001) & (euro_to_dollar['Time'].dt.year < 2021)]
bush = bush_obama_trump.copy(
       )[bush_obama_trump['Time'].dt.year < 2009]
obama = bush_obama_trump.copy(
       )[(bush_obama_trump['Time'].dt.year >= 2009) & (bush_obama_trump['Time'].dt.year < 2017)]
trump = bush_obama_trump.copy(
       )[(bush_obama_trump['Time'].dt.year >= 2017) & (bush_obama_trump['Time'].dt.year < 2021)]
```

To build the graphs the following approach will be used:

1. Build three of the graphs on a 2-by-3 grid (this grid should have six subplots, but we only build three; the bottom row remains empty).
2. Build only the bottom graph of a 2-by-1 grid (this grid should have two subplots; the top row remains empty).
3. The two grids are merged, and we end up with three graphs on the top row and one graph on the bottom row.

```python
import matplotlib.ticker as mticker

### Adding the FiveThirtyEight style
style.use('fivethirtyeight')

### Adding the subplots
plt.figure(figsize=(12, 6))
ax1 = plt.subplot(2,3,1)
ax2 = plt.subplot(2,3,2)
ax3 = plt.subplot(2,3,3)
ax4 = plt.subplot(2,1,2)
axes = [ax1, ax2, ax3, ax4]

### Changes to all the subplots
for ax in axes:
    ax.set_ylim(0.8, 1.7)
    ax.set_yticks([1.0, 1.2, 1.4, 1.6])
    ax.set_yticklabels(['1.0', '1.2','1.4', '1.6'],
                   alpha=0.3)
    ax.grid(alpha=0.5)    
    

### Ax1: Bush
ax1.plot(bush['Time'], bush['rolling_mean'],
        color='#BF5FFF')
ticks_loc = ax1.get_xticks().tolist()
ax1.xaxis.set_major_locator(mticker.FixedLocator(ticks_loc))
ax1.set_xticklabels(['', '2001', '', '2003', '', '2005', '',
                     '2007', '', '2009'],
                   alpha=0.3)
ax1.text(12149.0, 1.92, 'BUSH', fontsize=18, weight='bold',
        color='#BF5FFF')
ax1.text(11800.0, 1.8, '(2001-2009)', weight='bold',
        alpha=0.3)


### Ax2: Obama
ax2.plot(obama['Time'], obama['rolling_mean'],
        color='#ffa500')
ticks_loc = ax2.get_xticks().tolist()
ax2.xaxis.set_major_locator(mticker.FixedLocator(ticks_loc))
ax2.set_xticklabels(['', '2009', '', '2011', '', '2013', '',
                     '2015', '', '2017'],
                   alpha=0.3)
ax2.text(15149.0, 1.92, 'OBAMA', fontsize=18, weight='bold',
        color='#ffa500')
ax2.text(14900.0, 1.8, '(2009-2017)', weight='bold',
         alpha=0.3)


### Ax3: Trump
ax3.plot(trump['Time'], trump['rolling_mean'],
        color='#00B2EE')
ticks_loc = ax3.get_xticks().tolist()
ax3.xaxis.set_major_locator(mticker.FixedLocator(ticks_loc))
ax3.set_xticklabels(['2017', '', '2018', '', '2019', '',
                     '2020', '', '2021'],
                   alpha=0.3)
ax3.text(17610.0, 1.92, 'TRUMP', fontsize=18, weight='bold',
        color='#00B2EE')
ax3.text(17610.0, 1.8, '(2017-2021)', weight='bold',
         alpha=0.3)

### Ax4: Bush-Obama-Trump
ax4.plot(bush['Time'], bush['rolling_mean'],
        color='#BF5FFF')
ax4.plot(obama['Time'], obama['rolling_mean'],
        color='#ffa500')
ax4.plot(trump['Time'], trump['rolling_mean'],
        color='#00B2EE')
ax4.grid(alpha=0.5)
ax4.set_xticks([])

### Adding a title and a subtitle
ax1.text(10149.0, 2.14, '''EURO-USD exchange rates under George W. Bush (2001 - 2009), Barack Obama (2009-2017),
and Donald Trump (2017-2021)''',
          fontsize=20, weight='bold')

plt.show()
```

```python
import numpy as np

bush_max =  bush['US_dollar'].max()
obama_max = obama['US_dollar'].max()
trump_max = trump['US_dollar'].max()

data = {'George W. Bush (2001-2009)':bush_max,
        'Barack Obama (2009-2017)':obama_max,
        'Donald Trump (2017-2021)':trump_max}

group_data = list(data.values())
group_names = list(data.keys())

def addtext(x,y):
    for i in range(len(x)):
        plt.text(i,y[i]+ 0.03,y[i], ha = 'center',color='black')

        
# Main function

if __name__ == '__main__':
    
    # Define data
    x = group_names
    y = group_data

    fig, ax = plt.subplots(figsize=(8, 5.5))
    
    # Set background color of the plot
    ax.set_facecolor('#EFF7FF')


    ### Remove the x tick labels
    ax.set_yticklabels([])

    ### Remove the grid
    ax.grid(b=False)


    # Vertical bars
    ax.bar(group_names, group_data, alpha=0.4, width=0.1)

    ax.axhline(y=1.22, xmin=0.0, xmax=1.0, color='r', alpha= 0.4, linewidth = 1.0)
    
     # Call function
    addtext(x, y)   
    
    ax.text(-0.29, 1.22, '1.22',
        color='black', alpha=1)
    
    ### Adding a title and a subtitle
    ax.text(-0.5, 2.04, 'The average EURO-USD rate under the last three US presidents was 1.22',
             fontsize=20, weight='bold')
    ax.text(-0.5, 1.9, '''Highest EURO-USD exchange rates under the last three US presidents''',
            fontsize=16)
   
```
